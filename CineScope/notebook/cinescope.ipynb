2. Exploration data analysis

```python
#info tipe data dan missing value
print("=" * 50)
print("Info Dataset")
print("=" * 50)
df.info()
```

```python
from pandas._libs import missing
print("=" * 50)
print("Missing Value")
print("=" * 50)

missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
missing_df = pd.DataFrame({
    "Missing_Count" : missing,
    "Missing %" : missing_pct
}).query('Missing_Count > 0').sort_values('Missing %', ascending=False)

if missing_df.empty:
    print("Tidak ada missing value")
else:
    print(missing_df)

```

```python
print("=" * 50)
print('Statistik Deskriptif')
print("="*50)

df.describe().round(2)
```

```python
#cek distribusi imb rating

fig = px.box(
    df,
    x = 'decade',
    y = 'imdb_rating',
    color = 'decade',
    title = 'Distribusi IMDb Rating per Dekade',
    labels={'decade' : 'Dekade', 'imdb_rating' : 'IMDb Rating'},
    template='plotly_dark'
)

fig.update_layout(showlegend=False)
fig.show()
```

```python
#genre paling populer

#genre perlu dipecah karena format "Action|Drama|Sci-Fi"
genre_split = df['genre'].str.split('|').explode()
genre_counts = genre_split.value_counts().reset_index()
genre_counts.columns = ['genre', 'count']

fig = px.bar(
    genre_counts.head(15),
    x = 'count',
    y = 'genre',
    orientation='h',
    color = 'count',
    title = 'Top 15 genre populer',
    labels={'count' : 'Jumlah Film', 'genre' : 'Genre'},
    color_continuous_scale = 'viridis',
    template = 'plotly_dark'
    )

fig.update_layout(yaxis = {'categoryorder' : 'total ascending'})
fig.show()
```

```python
from plotly.graph_objs import XAxis
#Tren Budget vs Revenue per Tahun


yearly = df.groupby('release_year').agg(
    avg_budget = ('budget_million', 'mean'),
    avg_revenue = ('revenue_million', 'mean')
).reset_index()

fig = make_subplots(specs=[[{"secondary_y": False}]])

fig.add_trace(go.Scatter(
    x=yearly['release_year'], y=yearly['avg_budget'], name='Average Budget ($M)', line=dict(color='#f39c12', width=2)
))

fig.add_trace(go.Scatter(
    x = yearly['release_year'], y=yearly['avg_revenue'], name = 'Avg Revenue ($M)', line=dict(color = '#2ecc71', width=2)
))

fig.update_layout(
    title='Tren rata rata tahunan vs revenue per tahun',
    xaxis_title = 'Tahun',
    yaxis_title = 'Nilai ($ Juta)',
    template = 'plotly_dark',
    hovermode = 'x unified'
)

fig.show()
```
